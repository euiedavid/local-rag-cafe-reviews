```python
# check if review has empty strings
df['Review'].isna().sum()
```

```python
# check review column duplicates
df['Review'].duplicated().sum()
```

Duplicate review texts were identified. Most duplicates consist of generic short reviews such as "good" and occur across different restaurants, while one pair appears to be a duplicate record for the same restaurant. These records will be retained pending further validation.

```python
# inspect dups
# check all occurances
df[df['Review'].duplicated(keep=False)].sort_values('Review')
```

```python
# check unique values for overall rating
df['Overall_Rating'].unique()
```

```python
# count how many times each rating appears
df['Overall_Rating'].value_counts()
```

```python
# to include missing values
df['Overall_Rating'].value_counts(dropna=False)
```

```python
# check if rate for two is consistently formatted
df['Rate for two'].unique()
```

```python
# inspect if rate for two contains unexpected characters
df['Rate for two'].astype(str).str.extract(r'([^\d\s])')[0].value_counts(dropna=False)
```

```python
# check whether non-empty values consist only of digits
df['Rate for two'].astype(str).str.strip().str.fullmatch(r'\d+').value_counts()
```
